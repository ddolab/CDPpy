```python
# import required built-in functions from the package
from CDPpy.helper import input_path
from CDPpy import FedBatchCellCulture, FedBatchParameters

# Specify parameters that store data information for cell line 1 (CL1)
param_1 = FedBatchParameters(
          # input the cell line name, be consistent with the input Excel file
          cell_line_name='0',
          # input "True" if there are measurements on concentrations after feeding, otherwise, "False"
          use_concentration_after_feed=True,
          # input "True" if feeding composition is known, otherwise, "False"
          use_feed_concentration=False)

# Specify parameters that store data information for cell line 1 (CL1)
param_2 = FedBatchParameters(
          # input the cell line name, be consistent with the input Excel file
          cell_line_name='1',
          # input "True" if there are measurements on concentrations after feeding, otherwise, "False"
          use_concentration_after_feed=True,
          # input "True" if feeding composition is known, otherwise, "False"
          use_feed_concentration=False)

# Specify parameters that store data information for cell line 1 (CL1)
param_3 = FedBatchParameters(
          # input the cell line name, be consistent with the input Excel file
          cell_line_name='2',
          # input "True" if there are measurements on concentrations after feeding, otherwise, "False"
          use_concentration_after_feed=True,
          # input "True" if feeding composition is known, otherwise, "False"
          use_feed_concentration=False)



# check/printout the data processing setting
print(param_1)
print(param_2)
print(param_3)
```

```python
# create a Python object to perform the data processing of CL1
cell_line_1= FedBatchCellCulture()

# define path to the input data file. Change the filename if needed.
path = input_path('genentech_1200L_runs.xlsx')

# load the data set to the data-processing object
cell_line_1.load_data(file=path)

# perform data processing with specified data information.
cell_line_1.perform_data_process(parameters=[param_1, param_2, param_3])

# Save analysis output in excel file
cell_line_1.save_excel(file_name='output_genentech.xlsx')
```
